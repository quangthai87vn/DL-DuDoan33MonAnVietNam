```python
# Cell 3: Train YOLO11-seg trên dataset món ăn

from ultralytics import YOLO

# 1) Load model segmentation pretrained COCO
model = YOLO("yolo11s-seg.pt")  # hoặc yolo11m-seg.pt nếu GPU khoẻ

# 2) Đường dẫn yaml
#data_yaml = yaml_path  # hoặc "/Jupyter/datasets/mtl_food_seg/mtl_food_seg.yaml"
data_yaml ="/Users/admin/Documents/ai-dataset/vietnamese-foods/dataset/mtl_food_seg.yaml"
# 3) Train
results = model.train(
    data=data_yaml,
    epochs=100,        # tuỳ ông, 50–100 epoch là ổn
    imgsz=640,         # kích thước input
    batch=8,           # chỉnh theo VRAM (RTX 5000 16GB thì 16–32 cũng được)
    device=0,          # 0: GPU đầu tiên, -1: CPU
    workers=4,         # số luồng dataloader
    project="runs_seg",
    name="mtl_food_seg_01",
    exist_ok=True,     # cho phép ghi đè nếu đã tồn tại
    pretrained=True    # fine-tune từ COCO
)
```
