```python
import tensorflow as tf
import numpy as np

# Path to the TensorFlow Lite model
model_path = "yolo_models/food_yolov11_best_saved_model/food_yolov11_best_float32.tflite"

# Load TensorFlow Lite model
interpreter = tf.lite.Interpreter(model_path=model_path)
interpreter.allocate_tensors()

# Get input and output details
input_details = interpreter.get_input_details()
output_details = interpreter.get_output_details()

# Check input size (shape)
print("Input details:", input_details)
print("Output details:", output_details)

# Create dummy data to test the model
dummy_input = np.random.rand(1, 640, 640, 3).astype(np.float32)  # Assume the input is a 640x640 RGB image
interpreter.set_tensor(input_details[0]['index'], dummy_input)

# Perform inference
interpreter.invoke()

# Retrieve output results
output_data = interpreter.get_tensor(output_details[0]['index'])
print("Model output:", output_data)
```

```python
import torch
import coremltools as ct
from models.efficientnet_b0  import MTLEfficientNetB0   # file model của bạn

# Load BEST checkpoint để đánh giá
best_ckpt_path = "/Users/admin/Documents/DL-DuDoan33MonAnVietNam/Jupyter/runs/MTL_TGFOOD_01/checkpoints/mtl_effcientnet_b0_best.pt"
MAP_LOCATION="cpu"
ckpt = torch.load(best_ckpt_path, map_location=MAP_LOCATION)
print("Checkpoint keys:", ckpt.keys())

# Lấy đúng phần weight
state_dict = ckpt["model_state"]           # <<< quan trọng
class_names = ckpt.get("class_names", None)
print("Unexpected keys:", class_names)
model = MTLEfficientNetB0(num_classes=33)
missing, unexpected = model.load_state_dict(state_dict, strict=False)
print("Missing keys:", missing)
print("Unexpected keys:", unexpected)
model.eval()

# 2. Dummy input
example_input = torch.randn(1, 3, 224, 224)

# 3. Trace / convert
traced_model = torch.jit.trace(model, example_input)

# 4. Labels (33 món ăn)
class_labels = [
    "Banh beo",
    "Banh bot loc",
    "Banh can",
    "Banh canh",
    "Banh chung",
    "Banh cuon",
    "Banh duc",
    "Banh gio",
    "Banh khot",
    "Banh mi",
    "Banh pia",
    "Banh tet",
    "Banh trang nuong",
    "Banh xeo",
    "Bun bo Hue",
    "Bun dau mam tom",
    "Bun mam",
    "Bun rieu",
    "Bun thit nuong",
    "Ca kho to",
    "Canh chua",
    "Cao lau",
    "Chao long",
    "Com tam",
    "Goi cuon",
    "Hu tieu",
    "Mi quang",
    "Nem chua",
    "Pho",
    "Xoi xeo",
    "banh_da_lon",
    "banh_tieu",
    "banh_trung_thu"
]

mlmodel = ct.convert(
    traced_model,
    inputs=[ct.ImageType(name="input", shape=example_input.shape,
                         scale=1/255.0, bias=[0,0,0])],
    classifier_config=ct.ClassifierConfig(class_labels)
)




mlmodel.save("yolo_models/FoodMainDishEfficientNet.mlpackage")

```
